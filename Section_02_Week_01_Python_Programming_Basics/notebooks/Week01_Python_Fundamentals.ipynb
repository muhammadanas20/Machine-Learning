## Day 1 — Variables, Data Types, and Printing

**What it covers:** Python's core data types (int, float, string, list, tuple, dict, bool) and basic output.

**Why it matters for AI work:** Every dataset you touch is ultimately built from these types — a CSV row is a mix of ints/floats/strings, a model's config is a dict, a batch of training examples is a list (or a NumPy array, which is the "supercharged list" you'll meet in Week 2). If these aren't automatic, everything downstream feels harder than it needs to be.

**When to use what:**
- `list` — ordered, changeable (mutable) → use for sequences you'll modify (e.g. collecting results in a loop)
- `tuple` — ordered, unchangeable (immutable) → use for fixed groupings (e.g. a coordinate `(x, y)`) or dict keys
- `dict` — key → value lookup → use for anything you'd otherwise search a list for (fast lookups, config, labeled data)
- `bool` — flags/conditions

**Pitfalls:**
- Confusing `list` (mutable, `[]`) with `tuple` (immutable, `()`) — if you try `tuple_var[0] = 5` it will throw a `TypeError`. This is intentional, not a bug.
- Forgetting that `dict_var["missing_key"]` throws a `KeyError` — use `.get("key", default)` when the key might not exist.

## Day 2 — Control Flow: Conditionals, Loops, Functions as a Menu System

**What it covers:** `if/elif/else`, `for`/`while` loops, `break`/`continue`, and wrapping logic in functions.

**Why it matters for AI work:** Training loops (`for epoch in range(num_epochs):`), early stopping (`break` when validation loss stops improving), and data filtering (`if row["age"] > 18:`) are all this exact pattern, just applied to models instead of numbers.

**When to use what:**
- `for` loop — you know how many times / what you're iterating over (a list, a range, a file's lines)
- `while` loop — you're looping until a *condition* changes (e.g. "until the user quits", "until loss converges")
- `continue` — skip this iteration, keep looping
- `break` — stop the loop entirely

**Pitfalls:**
- `while True:` without a `break` path is an infinite loop — always make sure there's an exit condition.
- The `for...else` pattern (used in the prime-checker below) is unusual: the `else` runs only if the loop completed *without* hitting `break`. Easy to forget this exists, easy to misuse if you do remember it.
- In Python, `range(2, n)` stops **before** `n`, not at it — off-by-one errors are common here.


```python
# --- Conditionals: basic and nested ---
num = -50
if num > 0:
    print("Positive Number")
elif num == 0:
    print("Zero")
else:
    print("Negative Number")

age = 25
if age > 18:
    if age < 30:
        print("Young Adult")
    else:
        print("Adult")
```

```python
# --- Loops: continue, break, while, for ---
print("Skip even numbers:")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)

print("\nStop at 5:")
for i in range(10):
    if i == 5:
        break
    print(i)

print("\nCountdown with while:")
count = 5
while count > 0:
    print(count)
    count -= 1

print("\nLoop through a list:")
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# --- Exercise: Prime number checker (uses for...else) ---
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

for n in [1, 2, 15, 17, 29]:
    label = "is a prime number" if is_prime(n) else "is not a prime number"
    print(f"{n} {label}")
```

```python
# --- Exercise: Calculator menu (function dispatch pattern) ---
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b != 0:
        return a / b
    else:
        return "Division by zero is not allowed"

# In the original script this menu ran with input() in an infinite loop.
# For a notebook, we simulate a couple of "menu choices" directly so the
# cell runs top-to-bottom without blocking on user input.
operations = {"1": add, "2": subtract, "3": multiply, "4": divide}
sample_calls = [("1", 4, 5), ("4", 10, 0)]

for choice, a, b in sample_calls:
    result = operations[choice](a, b)
    print(f"Choice {choice}: {result}")
```

**Note on `input()`:** the original exercise used `input()` to build an interactive menu. That's the right approach in a `.py` script run from a terminal, but it blocks execution in a notebook (it waits forever for typed input). The pattern above (a dict mapping menu choice → function) is exactly what real command handlers / API routers look like — worth knowing regardless of `input()`.

## Day 3 — Functions, Scope, Recursion, and Custom Modules

**What it covers:** Function scope (local vs global), recursion, and splitting code across files (modules) with `import`.

**Why it matters for AI work:** Every ML project is organized as modules — a `data_loader.py`, a `model.py`, a `train.py` that imports from both. Understanding scope also prevents a very common bug class: accidentally overwriting a variable inside a function and being confused why the outer value didn't change.

**When to use what:**
- Keep a function's variables **local** by default — only reach for `global` when you have a specific, justified reason (rare in practice, and a common source of bugs when overused).
- Recursion — natural fit for tree/graph-like problems (parsing, search). For simple counting/looping, a plain loop is usually clearer and avoids Python's recursion depth limit (~1000 by default).
- Custom modules (`import my_module`) — use as soon as a script passes ~100–150 lines, or as soon as you're reusing the same functions in two places.

**Pitfalls:**
- Reading a variable from an *enclosing* scope works fine; **assigning** to it inside a function creates a new local variable instead, unless you declare `global varname`. This trips people up constantly.
- Recursive functions need a correct **base case** — `factorial(n)` below stops at `n == 0 or n == 1`. Miss that and you get infinite recursion → `RecursionError`.
- For `import math_operations as mo` to work outside a notebook, `math_operations.py` must be in the same folder (or on the Python path). Below we define it inline since a notebook is a single file.


```python
import math as m
print("Square root of 16:", m.sqrt(16))

# --- Global vs local scope ---
greeting = "Hi"

def say_hello():
    # reads the global 'greeting' fine (no assignment to it happening here)
    print(greeting + " from inside the function")

say_hello()
print(greeting + " from outside the function")

def greet():
    message = "Hello World"   # local variable, does not exist outside this function
    print(message)

greet()
# print(message)  # <- would raise NameError: 'message' only exists inside greet()
```

```python
# --- Recursion: factorial ---
def factorial(n):
    if n == 0 or n == 1:   # base case - without this, infinite recursion
        return 1
    else:
        return n * factorial(n - 1)

def print_factorial(n):
    result = factorial(n)
    print(f"The factorial of {n} is {result}")

print_factorial(5)
```

```python
# --- Custom module pattern ---
# In the original project this lived in its own file: math_operations.py
# and was imported as: import math_operations as mo
# Reproduced here as a class-like namespace so the notebook is self-contained.

class math_operations:
    @staticmethod
    def add(a, b):
        return a + b
    @staticmethod
    def subtract(a, b):
        return a - b
    @staticmethod
    def multiply(a, b):
        return a * b
    @staticmethod
    def divide(a, b):
        if b != 0:
            return a / b
        return "Division by zero is not allowed"

mo = math_operations
num1, num2 = 10, 5
print("Addition: ", mo.add(num1, num2))
print("Subtraction: ", mo.subtract(num1, num2))
print("Multiplication: ", mo.multiply(num1, num2))
print("Division: ", mo.divide(num1, num2))
```

## Day 4 — Data Structures: Lists, Dicts, Tuples, Sets

**What it covers:** The four built-in collection types and their core operations (add/remove/update/lookup).

**Why it matters for AI work:** This is direct groundwork for Pandas (Week 2). A Pandas DataFrame is essentially a dict of columns, where each column behaves like a list. Sets show up constantly for deduplication (e.g. unique labels in a classification dataset, unique words in an NLP vocabulary).

**When to use what:**
- `list` — ordered, allows duplicates, indexable (`fruits[0]`)
- `dict` — key → value, no duplicate keys, fast lookup by key
- `tuple` — like a list but immutable → safe to use as a dict key or in a set
- `set` — unordered, no duplicates → fast membership checks (`x in my_set` is much faster than `x in my_list` for large data) and set algebra (union, intersection, difference)

**Pitfalls:**
- Lists are **not** safe to use as dict keys (they're mutable) — you'll get a `TypeError: unhashable type: 'list'`. Use a tuple instead.
- `del my_dict["key"]` raises `KeyError` if the key doesn't exist; `my_dict.pop("key", None)` doesn't.
- Negative indexing (`fruits[-1]`) is a list-only convenience — sets have no order, so there's no `my_set[-1]`.


```python
# --- Dictionaries ---
person = {"name": "Alice", "age": 25, "grade": "A"}
print(person)

person["address"] = "123 Main St"   # add new key
person["age"] = 32                   # update existing key

if "grade" in person:
    del person["grade"]              # remove a key safely (checked first)

print(person)

for key, value in person.items():
    print(key, "->", value)
```

```python
# --- Lists ---
numbers = [1, 2, 3, 4]
fruits = ["apple", "banana", "cherry"]
mixed = [1, "apple", True]   # lists can hold mixed types (rarely a good idea in real code)

print(numbers[2])
print(fruits[-1])

fruits.append("orange")
fruits.insert(1, "grape")
print(fruits)

sliced_fruits = fruits[2:4]   # slicing: [start:stop), stop is excluded
print(sliced_fruits)

fruits.remove("banana")   # removes by VALUE
print(fruits)

del fruits[0]              # removes by INDEX
print(fruits)

fruits.pop()                # removes and returns the LAST item
print(fruits)
```

```python
# --- Tuples ---
colors = ("red", "green", "blue")
single_item = ("glass",)   # trailing comma required, or Python reads it as a plain string in parens

print(colors[0])
print(colors[-1])

# --- Sets ---
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print("Difference (set1 - set2):", set1 - set2)
print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
```

```python
# --- Exercise: Word frequency counter ---
def word_frequency(sentence):
    words = sentence.split()
    word_count = {}
    for word in words:
        word = word.lower()
        word_count[word] = word_count.get(word, 0) + 1   # .get() avoids a KeyError/if-check
    return word_count

sentence = "the quick brown fox jumps over the lazy dog the fox runs"
print(word_frequency(sentence))
```

**Why `.get(word, 0) + 1` matters:** this is the standard "counting dict" idiom. `collections.Counter` (from the standard library) does the same thing in one line — worth knowing once you've understood the manual version.

## Day 5 — Strings and Regular Expressions (Regex)

**What it covers:** String methods (`split`, `join`, `replace`, `strip`) and pattern matching with the `re` module.

**Why it matters for AI work:** This is the backbone of text preprocessing — the step before *any* NLP model (Week 12+) sees your data. Cleaning messy text (extra whitespace, punctuation, mixed case) directly affects model quality; garbage in, garbage out.

**When to use what:**
- String methods (`.split()`, `.replace()`, `.strip()`) — simple, fixed patterns (split on spaces, remove known characters)
- Regex (`re` module) — flexible patterns you can't hardcode (any digit sequence, any word boundary, "anything that looks like an email")

**Common regex patterns:**
- `\d+` — one or more digits
- `\w+` — one or more "word" characters (letters, digits, underscore)
- `\s+` — one or more whitespace characters
- `[^\w\s]` — anything that is NOT a word character or whitespace (i.e. punctuation)

**Pitfalls:**
- Regex is slower than plain string methods — don't reach for `re` when `.replace()` would do.
- Forgetting the `r` prefix (`r"\d+"` vs `"\d+"`) causes subtle bugs, because `\d` isn't a recognized Python escape sequence and gets treated ambiguously. Always use raw strings (`r"..."`) for regex patterns.
- `re.sub()` replaces ALL matches by default — pass a `count=` argument if you only want to replace the first N.


```python
import re

# --- Basic string methods ---
sentence = "Python is fun"
words = sentence.split()
print(words)

new_sentence = "|".join(words)
print(new_sentence)

text = "I love Java"
updated_text = text.replace("Java", "Python")
print(updated_text)

messy = "     Hello, World     "
print(repr(messy))
cleaned_text = messy.strip()
print(repr(cleaned_text))
```

```python
# --- Regex basics ---
text = "Contact me at 123-456-7890"

digits = re.findall(r"\d+", text)      # find all digit sequences
print("Digits found:", digits)

masked_text = re.sub(r"\d", "X", text)  # replace every digit with X
print("Masked:", masked_text)
```

```python
# --- Exercise: Text cleaning pipeline (typical NLP preprocessing step) ---
def clean_text(text):
    text = re.sub(r"[^\w\s]", "", text)   # strip punctuation
    text = " ".join(text.split())         # collapse extra whitespace
    return text.lower()                    # normalize case

input_text = "   Hello, World.!!! Welcome to Python, Programming....    "
print("Cleaned Text:", clean_text(input_text))
```

```python
# --- Exercise: Palindrome checker ---
def is_palindrome(text):
    text = "".join(char.lower() for char in text if char.isalnum())
    return text == text[::-1]

for sample in ["racecar", "A man a plan a canal Panama", "hello"]:
    verdict = "is" if is_palindrome(sample) else "is not"
    print(f'"{sample}" {verdict} a palindrome.')
```

## Day 6 — File I/O and Exception Handling

**What it covers:** Reading/writing text files with `open()`, and catching errors with `try/except`.

**Why it matters for AI work:** Every dataset starts as a file on disk. And every ML pipeline that reads external files (CSV, model weights, config, logs) needs to fail gracefully instead of crashing the whole run when a file is missing or corrupted.

**When to use what:**
- `with open(...) as f:` — always use this form. It auto-closes the file even if an error happens mid-read. Manually calling `f.close()` is easy to forget, especially after an exception.
- `try/except FileNotFoundError` — catch *specific* exceptions, not a bare `except:` (which silently swallows everything, including bugs you actually want to see, like `KeyboardInterrupt` or typos).

**Pitfalls:**
- Opening a file with mode `"w"` **overwrites** the entire file if it already exists. Use `"a"` (append) if you want to add to it instead.
- A bare `except:` hides real bugs — always name the exception type you expect (`FileNotFoundError`, `ValueError`, etc.).
- Forgetting `.strip()` when reading lines leaves trailing `\n` characters in your strings, which causes silent bugs later (e.g. `"apple\n" != "apple"`).


```python
# Create the sample files this section reads (so the notebook is self-contained)
with open("sample.txt", "w") as f:
    f.write("Hello World\nHow are you\nWhat is the plan for Today\nAre you working on AI Technologies\n")
```

```python
# --- Basic read with error handling ---
try:
    with open("sample.txt", "r") as file:
        content = file.read()
    print(content)
except FileNotFoundError:
    print("File Not Found!")
```

```python
# --- Exercise: Count words and lines in a file ---
def count_words_and_lines(filename):
    try:
        with open(filename, "r") as file:
            lines = file.readlines()
            line_count = len(lines)
            word_count = sum(len(line.split()) for line in lines)
            print(f"Number of lines: {line_count}")
            print(f"Number of words: {word_count}")
    except FileNotFoundError:
        print(f"File {filename} not found!")

count_words_and_lines("sample.txt")
```

```python
# --- Exercise: Write then read items from a file ---
def write_items_to_file(filename, items):
    with open(filename, "w") as file:   # "w" OVERWRITES the file - be careful
        for item in items:
            file.write(item + "\n")

def read_items_from_file(filename):
    try:
        with open(filename, "r") as file:
            items = file.readlines()
            print("Items in the file:")
            for item in items:
                print(item.strip())   # .strip() removes the trailing newline
    except FileNotFoundError:
        print(f"File {filename} not found!")

fruits = ["Apple", "Banana", "Cherry", "Dates"]
write_items_to_file("fruits.txt", fruits)
read_items_from_file("fruits.txt")
```

## Day 7 — Mini Project: Command-Line Task Manager

**What it covers:** Combining everything from the week — dicts, functions, file I/O, exception-safe parsing, and a menu loop — into one working program.

**Why it matters for AI work:** This is a "state management" pattern: load state from disk → mutate it in memory → save state back to disk. That's the same shape as checkpointing a model (load weights → train → save weights) or managing an experiment log.

**Design pattern used:** persistence functions (`load_tasks`, `save_tasks`) are separated from logic functions (`add_task`, `view_tasks`, ...). This separation — keeping "how state is stored" separate from "what the program does with it" — is a habit worth keeping. It means you could swap the storage from a text file to a database later without touching the logic functions.

**Pitfalls:**
- `task.strip().split(" | ")` (the original file, note the accidental space before `.split`) assumes every stored line has **exactly** the `id | title | status` format. If a task title itself contains `" | "`, parsing breaks. Real projects would use `csv`, `json`, or a database instead of a hand-rolled delimiter format for exactly this reason.
- `max(tasks.keys(), default=0) + 1` for generating new IDs works, but isn't safe if tasks can be deleted and re-added concurrently (two processes could compute the same "next ID"). Fine for a single-user CLI tool, not fine for anything multi-user.


```python
import os

FILE_NAME = "tasks.txt"

def load_tasks():
    tasks = {}
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            for line in file:
                line = line.strip()
                if not line:
                    continue
                task_id, title, status = line.split(" | ")
                tasks[int(task_id)] = {"title": title, "status": status}
    return tasks

def save_tasks(tasks):
    with open(FILE_NAME, "w") as file:
        for task_id, task in tasks.items():
            file.write(f"{task_id} | {task['title']} | {task['status']}\n")

def add_task(tasks, title):
    task_id = max(tasks.keys(), default=0) + 1
    tasks[task_id] = {"title": title, "status": "incomplete"}
    print(f"Task '{title}' added.")
    return task_id

def view_tasks(tasks):
    if not tasks:
        print("No tasks available.")
    else:
        for task_id, task in tasks.items():
            print(f"[{task_id}] {task['title']} - {task['status']}")

def mark_task_complete(tasks, task_id):
    if task_id in tasks:
        tasks[task_id]["status"] = "complete"
        print(f"Task '{tasks[task_id]['title']}' marked as complete.")
    else:
        print("Task ID not found.")

def delete_task(tasks, task_id):
    if task_id in tasks:
        deleted_task = tasks.pop(task_id)
        print(f"Task '{deleted_task['title']}' deleted.")
    else:
        print("Task ID not found.")
```

```python
# --- Simulated menu session (replaces the input()-driven while-loop menu) ---
# The original script wraps this all in `while True: choice = input(...)`.
# Below we call the same functions directly in sequence so it runs unattended.

tasks = load_tasks()

add_task(tasks, "Finish Week 1 notebook")
add_task(tasks, "Start Week 2 - NumPy")
view_tasks(tasks)

mark_task_complete(tasks, 1)
view_tasks(tasks)

delete_task(tasks, 2)
view_tasks(tasks)

save_tasks(tasks)
print("\nSaved to", FILE_NAME)
```
